```python
import numpy as np
scores= np.array([20,30,40,50,60,70,80,90,100])
student= np.array(["student 1","student 2","student 3","student 4","student 5","student 6","student 7","student 8","student 9"])
print("scores array:",scores)
```

```python
#How many scores are in the array#
print("there are",np.size(scores), "scores")
```

```python
#what is the highest score#
print("the highest score is",np.max(scores))
```

```python
#what is the lowest score#
print("the lowest score is", np.min(scores))
```

```python
#what is the average score#
print("the averag score is", np.average(scores))
```

```python
#what is the total score#
print("the total score is", np.sum(scores))
```

```python
#which student score above 50#
print("the students that score above 50 is", np.sum(scores>50))
```

```python
#which student score above 50#
print("the students that score above 50 is", student[scores>50])
```

```python
#which student score 50 or above#
print("the students that score 50 or above is", student[(scores>=50)])
```
